# Hash Tables and Hash Functions

**Goal**: Want to produce integers in the range of the table indices where they are well distributed ("randomly" distributed).

## Three Classic Techniques

1. Division Method
2. Multiplication Method
3. "MAD" Method (**M**ultiply-**A**dd-**D**ivide)

### Division Method

_Most Common_

$h(k) = k \mod m$

where $m$ is the size of the hash table.

### Multiplication Method

**Reason to Use**: Divisions (like `%`) are relatively slow.

$h(k) = \lfloor m(kA - \lfloor k * A \rfloor) \rfloor$

where

- $A$ is a constant in the range $0 < A < 1$.
- $m$ is the table size.

### MAD Method

$h(k) = | ak + b | \mod m$

where

- $a, b \in \mathbb{N}$
- $a, b, m$ are relatively prime


## Collision Resolution

Two basic collision resolution techniques:

1. Open Addressing
1. Chaining

### Open Addressing

**Idea**: When a collision occurs, we look for the next open slot.

### Chaining

**Idea**: Each slot in the hash table is a pointer to a linked list. When a collision occurs, we add the new element to the linked list.




```python
A = (5 ** 0.5 - 1) / 2
m = 1000
a, b = 34, 56
```

```python

def h_mod(k): return k % m
def h_mul(k): return int(m * ((k * A) % 1))
def h_mad(k): return abs(a * k + b) % m
```

```python
def linear_probe(k, i): 
  return (h_mod(k) + i) % m
```

```python
c1, c2 = 1, 1
def quadratic_probe(k, i): 
  return (h_mod(k) + c1 * i + c2 * i ** 2) % m
```

```python
def double_hash(k, i): 
  return (h_mod(k) + i * h_mul(k)) % m
```
